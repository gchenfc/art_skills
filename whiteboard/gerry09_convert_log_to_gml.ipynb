```python
def to_txy(log):
    codes = ['M', 'L', 'U']
    tptxy = np.loadtxt(log, delimiter=',', converters={1: lambda x: codes.index(x.decode())})
    txy = tptxy[:, 2:]
    penup = tptxy[:, 1]
    penup = (penup == codes.index('U'))

    # Cut up the stroke
    strokes = []
    cur = 0
    for ind in np.argwhere(penup).flatten() + 1:
        strokes.append(txy[cur:ind])
        cur = ind
    return strokes

def plot_txy(ax, strokes):
    # Plot the strokes
    for stroke in strokes:
        ax.plot(stroke[:, 1], stroke[:, 2])
        print(stroke.shape)
    ax.axis('equal');

def export(txys, name):
    with open(outfolder / f'{name}.json', 'w') as f:
        f.write(load_gml.txy_to_gml_json(txys, [0, 1, 0, 1]))
```
